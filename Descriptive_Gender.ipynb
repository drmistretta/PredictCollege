```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file into a DataFrame
file_path = '/content/drive/Shared drives/Dr. Sharon Mistretta/Add Health/Predict_College/data_with_birth_year.csv'
df = pd.read_csv(file_path)

print("Number of Columns in the dataset:", len(df.columns))

# Create a pie chart showing the percentage and count of male and female students
# Assuming H5OD2A is coded as 1 = Male and 2 = Female

gender_counts = df['H5OD2A'].value_counts().sort_index()  # Sort by index (1 for male, 2 for female)
total_count = gender_counts.sum()

# Define labels with both count and percentage
labels = [f'Male ({gender_counts[1]} - {gender_counts[1]/total_count:.1%})',
          f'Female ({gender_counts[2]} - {gender_counts[2]/total_count:.1%})']

# Plot the pie chart
colors = ['#e0ffff', '#b2d8d8']  # Customize colors as needed
plt.pie(gender_counts, labels=labels, autopct='%1.1f%%', startangle=90, colors=colors,
        wedgeprops={'edgecolor': 'black', 'linewidth': 2}, # Add border to wedge slices
        labeldistance=1.1,  # Adjust distance of labels from the center
        pctdistance=0.5)   # Adjust percentage placement

# Set the title with total count
plt.title(f'Percentage and Count of Male and Female Students (Total: {total_count})')

# Ensure the pie chart has an equal aspect ratio
plt.axis('equal')

# Display the pie chart
plt.show()
```
